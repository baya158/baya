```markdown
### ✂️ Division des Données (Train-Test Split)

Nous allons diviser nos données en utilisant la bibliothèque `scikit-learn` :
```

````python
from sklearn.model_selection import train_test_split

# Division des données : 75% train et 25% test
# random_state=42 permet d'obtenir des résultats reproductibles
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

print("Division effectuée avec succès !")
print(f"Taille de l'ensemble d'entraînement (X_train) : {X_train.shape} (75%)")
print(f"Taille de l'ensemble de test (X_test) : {X_test.shape} (25%)")
````

````python
print(f"Nombre de lignes pour l'entraînement (75%) : {X_train.shape[0]}")
print(f"Nombre de lignes pour la vérification finale (25%) : {X_test.shape[0]}")
````

````python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# 1. Initialisation des modèles
lin_reg = LinearRegression()
rf_reg = RandomForestRegressor(n_estimators=100, random_state=42)

# 2. Entraînement des modèles sur l'ensemble d'entraînement
print("Entraînement de la Régression Linéaire...")
lin_reg.fit(X_train, y_train)

print("Entraînement de la Forêt Aléatoire (cela peut prendre quelques secondes)...")
rf_reg.fit(X_train, y_train)

# 3. Prédictions sur l'ensemble de vérification (test)
y_pred_lin = lin_reg.predict(X_test)
y_pred_rf = rf_reg.predict(X_test)

# 4. Évaluation des performances
rmse_lin = np.sqrt(mean_squared_error(y_test, y_pred_lin))
r2_lin = r2_score(y_test, y_pred_lin)

rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

# 5. Affichage des résultats
print("\n================ RÉSULTATS DES MODÈLES ================")
print(f"Régression Linéaire :")
print(f"  - Erreur moyenne (RMSE) : {rmse_lin:.4f} (en centaines de milliers de $)")
print(f"  - Score R² (précision)  : {r2_lin:.4f} (soit {r2_lin*100:.2f}% de la variance expliquée)")

print(f"\nForêt Aléatoire :")
print(f"  - Erreur moyenne (RMSE) : {rmse_rf:.4f} (en centaines de milliers de $)")
print(f"  - Score R² (précision)  : {r2_rf:.4f} (soit {r2_rf*100:.2f}% de la variance expliquée)")
````

### 🧠 Rôle des Algorithmes d'Apprentissage Automatique sélectionnés

Pour modéliser et prédire la valeur moyenne des logements (`MedHouseVal`), nous utilisons deux algorithmes complémentaires :

#### 1. La Régression Linéaire (`LinearRegression`)
* **Rôle :** Établir une relation linéaire directe entre les caractéristiques ($X$) et la cible ($y$). L'équation cherche à tracer la "droite" (ou l'hyperplan) qui passe au plus près de tous les points de données.
* **Avantages :** Très simple, extrêmement rapide à entraîner et facile à interpréter (chaque coefficient indique l'impact direct d'une variable sur le prix).
* **Limites :** Elle suppose que la relation est linéaire. Si le prix d'une maison varie de manière complexe ou non-linéaire avec l'âge ou la localisation, cet algorithme sera moins performant.

#### 2. La Forêt Aléatoire (`RandomForestRegressor`)
* **Rôle :** C'est un algorithme d'apprentissage d'ensemble basé sur une multitude d'**arbres de décision**. Chaque arbre vote pour une prédiction, et la moyenne de leurs prédictions donne le résultat final.
* **Avantages :** Excellent pour capturer les relations non-linéaires complexes et les interactions entre les variables (par exemple, la combinaison de la latitude, de la longitude et du revenu d'un coup). Il est robuste et donne souvent de très bons résultats d'emblée.
* **Limites :** Plus lourd en calculs et plus difficile à interpréter visuellement qu'une simple ligne de régression.

````python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# 1. Initialisation des modèles
lin_reg = LinearRegression()
rf_reg = RandomForestRegressor(n_estimators=100, random_state=42)

# 2. Entraînement des modèles sur l'ensemble de train
print("Début de l'entraînement...")
print("Entraînement de la Régression Linéaire en cours...")
lin_reg.fit(X_train, y_train)

print("Entraînement de la Forêt Aléatoire en cours (cela peut prendre un instant)...")
rf_reg.fit(X_train, y_train)
print("Entraînement terminé !\n")

# 3. Prédiction sur l'ensemble de validation
y_pred_lin = lin_reg.predict(X_test)
y_pred_rf = rf_reg.predict(X_test)

# 4. Évaluation
rmse_lin = np.sqrt(mean_squared_error(y_test, y_pred_lin))
r2_lin = r2_score(y_test, y_pred_lin)

rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

# 5. Affichage des résultats
print("================ COMPARAISON DES RÉSULTATS ================")
print(f"Régression Linéaire :")
print(f"  -> Erreur moyenne (RMSE) : {rmse_lin:.4f} (en centaines de milliers de $)")
print(f"  -> Score de précision R² : {r2_lin:.4f} (soit {r2_lin*100:.2f}% de la variance expliquée)")

print(f"\nForêt Aléatoire :")
print(f"  -> Erreur moyenne (RMSE) : {rmse_rf:.4f} (en centaines de milliers de $)")
print(f"  -> Score de précision R² : {r2_rf:.4f} (soit {r2_rf*100:.2f}% de la variance expliquée)")
````

### 📊 Évaluation Statistique Détaillée et Visualisation des Performances

Pour valider scientifiquement nos modèles, nous analysons :
1. **L'erreur quadratique moyenne (RMSE)** : Pénalise fortement les grandes erreurs.
2. **L'erreur absolue moyenne (MAE)** : Représente l'écart moyen absolu sur le prix.
3. **Le coefficient de détermination ($R^2$)** : La proportion de la variance expliquée par le modèle.
4. **Le $R^2$ ajusté** : Ajuste le $R^2$ en fonction du nombre de variables explicatives utilisées.

````python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def calculer_r2_ajuste(r2, n, p):
    """Calcule le R² ajusté : n = nb d'échantillons, p = nb de variables"""
    return 1 - (1 - r2) * (n - 1) / (n - p - 1)

# Dimensions pour le calcul du R² ajusté
n_samples = X_test.shape[0]
p_features = X_test.shape[1]

# --- 1. Calcul des métriques pour la Régression Linéaire ---
mae_lin = mean_absolute_error(y_test, y_pred_lin)
rmse_lin = np.sqrt(mean_squared_error(y_test, y_pred_lin))
r2_lin = r2_score(y_test, y_pred_lin)
r2_adj_lin = calculer_r2_ajuste(r2_lin, n_samples, p_features)

# --- 2. Calcul des métriques pour la Forêt Aléatoire ---
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)
r2_adj_rf = calculer_r2_ajuste(r2_rf, n_samples, p_features)

# --- 3. Affichage des rapports statistiques ---
print("================ RAPPORT STATISTIQUE DÉTAILLÉ ================")
print(f"{ 'Métrique':<30} | { 'Régression Linéaire':<20} | { 'Forêt Aléatoire':<20}")
print("-" * 78)
print(f"{ 'Erreur Absolue Moyenne (MAE)':<30} | {f'{mae_lin:.4f}':<20} | {f'{mae_rf:.4f}':<20}")
print(f"{ 'Erreur Quadratique (RMSE)':<30} | {f'{rmse_lin:.4f}':<20} | {f'{rmse_rf:.4f}':<20}")
print(f"{ 'Coefficient R²':<30} | {f'{r2_lin:.4f}':<20} | {f'{r2_rf:.4f}':<20}")
print(f"{ 'R² Ajusté':<30} | {f'{r2_adj_lin:.4f}':<20} | {f'{r2_adj_rf:.4f}':<20}")
print("==============================================================\n")

# --- 4. Visualisation : Valeurs Réelles vs Valeurs Prédites ---
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Graphique Régression Linéaire
sns.scatterplot(x=y_test, y=y_pred_lin, alpha=0.3, color='blue', ax=axes[0])
axes[0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], '--r', lw=2)
axes[0].set_title(f"Régression Linéaire\n(R² = {r2_lin:.3f} | RMSE = {rmse_lin:.3f})")
axes[0].set_xlabel("Valeurs Réelles")
axes[0].set_ylabel("Prédictions")

# Graphique Forêt Aléatoire
sns.scatterplot(x=y_test, y=y_pred_rf, alpha=0.3, color='green', ax=axes[1])
axes[1].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], '--r', lw=2)
axes[1].set_title(f"Forêt Aléatoire\n(R² = {r2_rf:.3f} | RMSE = {rmse_rf:.3f})")
axes[1].set_xlabel("Valeurs Réelles")
axes[1].set_ylabel("Prédictions")

plt.tight_layout()
plt.show()
````

````python
from sklearn.metrics import r2_score
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Calcul de la précision moyenne en pourcentage (100% - MAPE)
# MAPE (Mean Absolute Percentage Error) représente le pourcentage d'écart moyen par rapport à la valeur réelle.
mape_lin = np.mean(np.abs((y_test - y_pred_lin) / y_test)) * 100
precision_lin = 100 - mape_lin

mape_rf = np.mean(np.abs((y_test - y_pred_rf) / y_test)) * 100
precision_rf = 100 - mape_rf

# Score R² exprimé en pourcentage
r2_pct_lin = r2_score(y_test, y_pred_lin) * 100
r2_pct_rf = r2_score(y_test, y_pred_rf) * 100

# 2. Affichage des décisions statistiques en pourcentage
print("================ DÉCISION ET EFFICACITÉ EN POURCENTAGE ================")
print(f"Régression Linéaire :")
print(f"  -> Capacité d'explication de la variance (R²) : {r2_pct_lin:.2f}%")
print(f"  -> Précision moyenne des prédictions : {precision_lin:.2f}%")
print(f"  -> Écart d'erreur moyen : {mape_lin:.2f}%")

print(f"\nForêt Aléatoire :")
print(f"  -> Capacité d'explication de la variance (R²) : {r2_pct_rf:.2f}%")
print(f"  -> Précision moyenne des prédictions : {precision_rf:.2f}%")
print(f"  -> Écart d'erreur moyen : {mape_rf:.2f}%")
print("========================================================================\n")

# 3. Visualisation claire des erreurs de prédiction en pourcentage
# On calcule les erreurs relatives individuelles en % pour chaque point de données
erreur_pct_lin = np.abs((y_test - y_pred_lin) / y_test) * 100
erreur_pct_rf = np.abs((y_test - y_pred_rf) / y_test) * 100

plt.figure(figsize=(14, 6))

# Distribution des erreurs en pourcentage
sns.kdeplot(erreur_pct_lin, fill=True, color="blue", label=f"Régression Linéaire (Erreur moy. : {mape_lin:.1f}%)", alpha=0.4)
sns.kdeplot(erreur_pct_rf, fill=True, color="green", label=f"Forêt Aléatoire (Erreur moy. : {mape_rf:.1f}%)", alpha=0.4)

plt.xlim(0, 100) # On focalise sur la zone de 0 à 100% d'erreur pour une meilleure lisibilité
plt.title("Comparaison de la distribution des erreurs de prédiction (en %)", fontsize=14)
plt.xlabel("Erreur de prédiction absolue (% d'écart par rapport au prix réel)", fontsize=12)
plt.ylabel("Densité des prédictions", fontsize=12)
plt.axvline(x=10, color='red', linestyle='--', alpha=0.7, label='Seuil de haute précision (10% d\'erreur)')
plt.legend(fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
````

### 🚀 Entraînement de la Régression Linéaire (`model_100`)

Voici les étapes détaillées du code :
1. **Importation** : Nous importons l'algorithme `LinearRegression` depuis la bibliothèque scikit-learn.
2. **Instanciation** : Nous créons notre instance de modèle nommée `model_100`.
3. **Entraînement** : Nous ajustons le modèle sur l'intégralité des données d'entraînement (`X_train`, `y_train`) via la méthode `.fit()`.
4. **Prédiction** : Nous générons les prédictions sur les données de test (`X_test`) sans les inclure dans l'entraînement, et stockons les résultats dans `y_pred_100`.
5. **Confirmation** : Un message s'affiche pour valider le succès de l'opération.

````python
from sklearn.linear_model import LinearRegression

# 1. Création de l'instance du modèle
model_100 = LinearRegression()

# 2. Entraînement sur 100% des données d'apprentissage disponibles
model_100.fit(X_train, y_train)

# 3. Prédiction sur l'ensemble de vérification (X_test)
y_pred_100 = model_100.predict(X_test)

print("Entraînement du modèle 'model_100' terminé avec succès et prédictions générées !")
````

### ⚖️ Comparaison : Entraînement sur 100% vs 50% des données d'apprentissage

Pour étudier l'impact du volume de données sur l'apprentissage :
1. **Extraction de 50% des données** : Nous divisons `X_train` et `y_train` avec un ratio de `0.5` pour obtenir exactement la moitié de l'échantillon d'apprentissage.
2. **Instanciation** : Création de `model_50` (Régression Linéaire).
3. **Apprentissage** : Entraînement sur la fraction restreinte (`X_train_50`, `y_train_50`).
4. **Prédiction** : Évaluation sur le **même** ensemble de test indépendant (`X_test`) stockée dans `y_pred_50`.
5. **Comparaison** : Calcul et affichage des scores R² et des erreurs moyennes pour comparer scientifiquement l'apport quantitatif des données.

````python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# 1. Sélectionner exactement 50% de l'ensemble d'entraînement initial
X_train_50, _, y_train_50, _ = train_test_split(
    X_train,
    y_train,
    train_size=0.5,
    random_state=42
)

# 2. Création et entraînement de model_50
model_50 = LinearRegression()
model_50.fit(X_train_50, y_train_50)

# 3. Prédictions sur le même ensemble de test (X_test)
y_pred_50 = model_50.predict(X_test)

# 4. Évaluation des deux modèles
rmse_100 = np.sqrt(mean_squared_error(y_test, y_pred_100))
r2_100 = r2_score(y_test, y_pred_100)

rmse_50 = np.sqrt(mean_squared_error(y_test, y_pred_50))
r2_50 = r2_score(y_test, y_pred_50)

# 5. Affichage comparatif
print(f"Taille d'entraînement pour model_100 : {X_train.shape[0]} lignes")
print(f"Taille d'entraînement pour model_50  : {X_train_50.shape[0]} lignes\n")
print("================ COMPARAISON DES DEUX MODÈLES ================")
print(f"Modèle 100% de données (model_100) :")
print(f"  - R² : {r2_100*100:.4f}%")
print(f"  - RMSE : {rmse_100:.4f}")
print(f"\nModèle 50% de données (model_50) :")
print(f"  - R² : {r2_50*100:.4f}%")
print(f"  - RMSE : {rmse_50:.4f}")
print("==============================================================")
````
